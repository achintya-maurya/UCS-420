Q1. Consider the following sensor readings:
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
A sensor has an average measurement error of +2°C.
Perform the following without using a for loop:
a. Add 2°C to every reading using vectorization.
b. Convert all temperatures from Celsius to Fahrenheit using:
F =
9
5
C + 32
c. Identify all readings greater than 32°C using Boolean indexing.
d. Count how many readings exceed 32°C.
e. Explain how vectorization and Boolean indexing can make data processing more efficient
than explicitly iterating through every value.

```python

import numpy as np

# Original sensor readings in Celsius
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# (a) Add 2°C to every reading using vectorization
corrected_temperature = temperature + 2

print("(a) Corrected temperatures (°C):")
print(corrected_temperature)

# (b) Convert corrected temperatures from Celsius to Fahrenheit
fahrenheit = (9 / 5) * corrected_temperature + 32

print("\n(b) Temperatures in Fahrenheit:")
print(fahrenheit)

# (c) Identify readings greater than 32°C using Boolean indexing
readings_above_32 = corrected_temperature[corrected_temperature > 32]

print("\n(c) Readings greater than 32°C:")
print(readings_above_32)

# (d) Count the readings greater than 32°C
count = np.sum(corrected_temperature > 32)

print("\n(d) Number of readings greater than 32°C:")
print(count)
```

Q2. The following matrix represents the number of daily steps recorded by four users
over three days:
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])
Perform the following:
a. Calculate the total steps recorded.
b. Calculate the mean number of steps.
c. Find the maximum and minimum values.
d. Calculate the total steps for each day using axis=0.
e. Calculate the total steps for each user using axis=1.
f. Find the user/day position containing the maximum number of steps using argmax().

```python

import numpy as np

# Daily steps recorded by four users over three days
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# (a) Calculate total steps recorded
total_steps = np.sum(steps)
print("(a) Total steps recorded:", total_steps)

# (b) Calculate mean number of steps
mean_steps = np.mean(steps)
print("(b) Mean number of steps:", mean_steps)

# (c) Find maximum and minimum values
maximum_steps = np.max(steps)
minimum_steps = np.min(steps)

print("(c) Maximum steps:", maximum_steps)
print("    Minimum steps:", minimum_steps)

# (d) Total steps for each day (axis=0)
daily_totals = np.sum(steps, axis=0)
print("(d) Total steps for each day:", daily_totals)

# (e) Total steps for each user (axis=1)
user_totals = np.sum(steps, axis=1)
print("(e) Total steps for each user:", user_totals)

# (f) Find the position of the maximum value using argmax()
max_position = np.unravel_index(np.argmax(steps), steps.shape)

print("(f) Maximum value position (row, column):", max_position)
print("    User number:", max_position[0] + 1)
print("    Day number:", max_position[1] + 1)
```

Q3. Write a Python program using NumPy to perform the following operations:
a. Create the following NumPy array:
original = np.array([1, 2, 3, 4, 5, 6])
b. Create a slice containing elements from index 1 to 4 and store it in a variable named subset.
c. Modify the first element of subset to 999 and display both original and subset.
d. Create another slice from original, but this time use .copy(). Modify its first element to 500
and display original and the copied array.
e. Create a NumPy array containing the numbers from 1 to 12 using np.arange() and reshape it
into a 3 × 4 matrix.
f. From the 3 × 4 matrix, extract the following using NumPy indexing and slicing:
• First row
• Last row
• Second column
• Elements from rows 1–2 and columns 2–3
g. Create a flattened version of the 3 × 4 matrix using both flatten() and ravel().
h. Modify an element in the array returned by ravel() and observe whether the original array is
affected.
i. Modify an element in the array returned by flatten() and observe whether the original array is
affected.
j. Display the shape, ndim, size, and dtype of the original 3 × 4 matrix.

```python

import numpy as np

# (a) Create the original NumPy array
original = np.array([1, 2, 3, 4, 5, 6])
print("(a) Original array:")
print(original)

# (b) Create a slice from index 1 to index 4 (exclusive)
subset = original[1:5]
print("\n(b) Subset:")
print(subset)

# (c) Modify the first element of the subset
subset[0] = 999
print("\n(c) After modifying subset:")
print("Original array:", original)
print("Subset:", subset)

# (d) Create an independent copy and modify it
copied_array = original[1:5].copy()
copied_array[0] = 500
print("\n(d) After modifying copied array:")
print("Original array:", original)
print("Copied array:", copied_array)

# (e) Create and reshape an array from 1 to 12
matrix = np.arange(1, 13).reshape(3, 4)
print("\n(e) 3 x 4 matrix:")
print(matrix)

# (f) Extract rows, columns, and a submatrix
print("\n(f) First row:")
print(matrix[0, :])

print("Last row:")
print(matrix[-1, :])

print("Second column:")
print(matrix[:, 1])

print("Rows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

# (g) Flatten using flatten() and ravel()
flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("\n(g) Flatten result:")
print(flat_array)

print("Ravel result:")
print(ravel_array)

# (h) Modify the array returned by ravel()
ravel_array[0] = 100
print("\n(h) After modifying ravel array:")
print("Ravel array:", ravel_array)
print("Original matrix:")
print(matrix)

# (i) Modify the array returned by flatten()
flat_array[1] = 200
print("\n(i) After modifying flatten array:")
print("Flatten array:", flat_array)
print("Original matrix:")
print(matrix)

# (j) Display matrix properties
print("\n(j) Matrix properties:")
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

Q4. A simple cognitive-assistive system wants to predict a user's assistance score based
on three factors:
• Sleep hours
• Activity level
• Stress level
Consider:
X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])
y = np.array([40, 65, 30, 85])
Here, each row represents one user and each column represents a feature.
Perform the following:
a. Display the shape and dimensions of X.
b. Calculate: X.T and explain what the transpose represents.
c. Calculate the matrix product: X.T @ X
d. Calculate the matrix inverse using: np.linalg.inv() where appropriate.
e. Implement the Ordinary Least Squares equation: 𝛽 = (𝑋!𝑋)"#𝑋!𝑦 using NumPy.
f. Explain what the resulting coefficients represent in the context of the three user
features.
g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to
obtain a predicted assistance score.

```python

import numpy as np

# Given data: Sleep hours, Activity level, Stress level
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# Assistance scores
y = np.array([40, 65, 30, 85])

# (a) Display shape and dimensions
print("(a) Shape of X:", X.shape)
print("    Dimensions of X:", X.ndim)

# (b) Calculate transpose of X
X_transpose = X.T
print("\n(b) Transpose of X:")
print(X_transpose)

# (c) Calculate X.T @ X
product = X_transpose @ X
print("\n(c) X.T @ X:")
print(product)

# (d) Calculate the inverse of X.T @ X
inverse = np.linalg.inv(product)
print("\n(d) Inverse of X.T @ X:")
print(inverse)

# (e) Calculate OLS coefficients
beta = inverse @ X_transpose @ y
print("\n(e) OLS coefficients:")
print(beta)

# (f) Display the meaning of the coefficients
print("\n(f) Coefficients:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

# (g) Predict assistance score for a new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\n(g) Predicted assistance score:", predicted_score)
```
