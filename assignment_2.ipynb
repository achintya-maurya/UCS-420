1. Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L (10
numbers in list now).

i. Print L

ii. Add two numbers of your choice to L — one using append() and one using insert() at a specific position.

Print L after each and explain in a comment what changed.

iii. Remove two elements from L — one using remove() and one using pop(). Print L after each.

iv. Sort L in ascending order using sort(), then descending order using sort(reverse=True). Print both.

v. Use slicing to print the first three elements and the last three elements of L in a single line each.

vi. Use list comprehension to create a new list containing only the elements of L that are greater than the
average of L.

```python

roll_number = "1024170429"   

# Extract digits and multiply each by 10
L = [int(digit) * 10 for digit in roll_number]

print("Roll Number:", roll_number)
print("List L:", L)

# i. Print L
print("L =", L)

# ii. append() and insert()
L.append(500)
# append() adds element at the END of list
print("After append(500):", L)

L.insert(3, 999)
# insert() adds element at given index; later elements shift right
print("After insert(3, 999):", L)

# iii. remove() and pop()
L.remove(999)
# remove() deletes the first matching VALUE from list
print("After remove(999):", L)

removed_element = L.pop(0)
# pop() deletes element at given INDEX and returns it
print("Removed element (via pop):", removed_element)
print("After pop(0):", L)

# iv. sort() ascending and descending
L.sort()
# sorts list in ascending order (in-place)
print("L in Ascending Order:", L)

L.sort(reverse=True)
# sorts list in descending order (in-place)
print("L in Descending Order:", L)

# v. Slicing - first 3 and last 3 elements
print("First 3 elements:", L[:3])
print("Last 3 elements:", L[-3:])

# vi. List comprehension - elements greater than average
average_L = sum(L) / len(L)
print("Average of L:", average_L)

greater_than_avg = [x for x in L if x > average_L]
# keep only elements greater than the average
print("Elements greater than average:", greater_than_avg)
```

3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating
your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed
produces a different list, so this cannot be copied from a classmate).

i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

ii. Count and print all odd numbers in the list.

iii. Count and print all even numbers in the list.

iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers

found using list comprehension.
v. Print the number that occurs most frequently in your list, and how many times it occurs

```python
import random

# Set seed to your roll number for reproducible but unique random list
random.seed(1234567890)   

# i. Generate 100 random numbers between 100 and 900 (inclusive)
random_list = [random.randint(100, 900) for _ in range(100)]
print("Random List:", random_list)

# ii. Count odd numbers
odd_numbers = [x for x in random_list if x % 2 != 0]
print("Odd numbers count:", len(odd_numbers))

# iii. Count even numbers
even_numbers = [x for x in random_list if x % 2 == 0]
print("Even numbers count:", len(even_numbers))

# iv. Count prime numbers
def is_prime(n):
    # check divisibility from 2 up to sqrt(n)
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_list if is_prime(x)]
print("Prime numbers count:", len(prime_numbers))
print("Prime numbers list:", prime_numbers)

# v. Most frequent number and its count
from collections import Counter

counts = Counter(random_list)
most_common_number, frequency = counts.most_common(1)[0]
print("Most frequent number:", most_common_number, "occurs", frequency, "time(s)")
```

4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
each digit} and set B = {digit × 9 for each digit}. Print both sets.
vi. Find the union of A and B — all unique values across both sets.
vii. Find the intersection of A and B — values common to both.
viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
differs from symmetric_difference().
ix. Find the symmetric difference of A and B.
x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().
xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain
in one line why discard() is safer than remove() when you're not sure the value exists

```python
# Using first 8 digits of roll number (before multiplying by 10)
digits = [int(d) for d in roll_number[:8]]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

# vi. Union - all unique values from both sets
union_AB = A.union(B)
print("Union (A ∪ B):", union_AB)

# vii. Intersection - values common to both
intersection_AB = A.intersection(B)
print("Intersection (A ∩ B):", intersection_AB)

# viii. Difference A-B and B-A
diff_A_B = A.difference(B)
diff_B_A = B.difference(A)
print("A - B:", diff_A_B)
print("B - A:", diff_B_A)
# difference() gives one-sided unique values, while symmetric_difference() gives both sides combined

# ix. Symmetric difference
sym_diff = A.symmetric_difference(B)
print("Symmetric Difference:", sym_diff)

# x. Subset and superset check
print("Is A subset of B?", A.issubset(B))
print("Is B superset of A?", B.issuperset(A))

# xi. Remove user input value from A using discard()
X = int(input("Enter a value to remove from set A: "))
A.discard(X)
# discard() does nothing if value is missing, while remove() raises an error - safer when unsure
print("Set A after discard:", A)
```

```python
my_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1234567890",
    "branch": "Computer Science",
    "age": 20,
    "city": "Patiala"
}

# i. Rename "city" to "location" using pop() (works for any dictionary)
my_dict["location"] = my_dict.pop("city")
print("After renaming city to location:", my_dict)

# ii. Add new key "cgpa"
my_dict["cgpa"] = 8.5
print("After adding cgpa:", my_dict)

# iii. Update age by increasing by 1
my_dict["age"] += 1
print("After updating age:", my_dict)

# iv. Delete "branch" using pop() and del in two separate copies
dict_copy1 = my_dict.copy()
removed_value = dict_copy1.pop("branch")
print("Using pop(), removed value:", removed_value)
print("Dict after pop():", dict_copy1)

dict_copy2 = my_dict.copy()
del dict_copy2["branch"]
print("Dict after del:", dict_copy2)
# pop() returns the removed value, while del does not return anything

# v. Iterate over dictionary using .items()
for key, value in my_dict.items():
    print(f"{key} → {value}")

# vi. Check for "email" key before accessing
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email not found in dictionary")

# vii. Merge with friend_dict
friend_dict = {
    "name": "Aman Verma",
    "roll_no": "9876543210",
    "branch": "Electronics",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:", merged_dict)
# When keys overlap, values from the second dictionary (friend_dict) overwrite the first

# viii. Dictionary comprehension - only string value pairs
string_value_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("Dictionary with only string values:", string_value_dict)
```
